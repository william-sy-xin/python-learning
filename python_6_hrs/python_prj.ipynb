```python
menu = {
    "burger" : 60,
    "popcorn" :  20,
    "Milk" : 10,
    "rice" : 30,
    "fries":8,
    "kids menu" : 50,
    "bottle of water" : 0
}
#drinks = {
    #"cola" = 5,
    #"cofee" = 10,
    #"sprite" = 5,
    #"lemonnade" = 8,
    #"iced tea" = 10
#}
print ("menu")
#print ("drinks")
print("---------------------")
total = 0
cart = []
for item, price in menu.items():
    print(f"{item}: {price}") 
while True:
    food = input ("please enter a menu item and (enter q to finsh)")
    if food == "q":
        break
    elif menu.get(food) is None:
        print ("order a food already!")
    else:
        cart.append(food)
        total += menu.get(food)
        print (food, end=" ")

print(f"your total: {total} dollars ") 
```
