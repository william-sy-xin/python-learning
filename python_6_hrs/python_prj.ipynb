```python
operator = input ('put in the equation (addition : + ,multiplucation : - , substraction : * division : /) : ') 
num1 = float(input('put in the first number in the equation : '))
num2 = float (input('put in the second numer in the equation : '))

if operator == '+' :
    result = num1 + num2
    print ({result})

elif operator == '-' :
    result = num1 - num2

elif operator == '*' :
    result = num1 * num2

elif operator == '/' :
    result = num1 / num2

else : 
    print ("symbol not regconized")

print (f"The result of problem is {result}")
```

```python
weight = float(input('put in the weight:'))
unit = input("the weight of the object is kilogram or pounds? (kg/lb)").upper ()

if unit  == 'KG' :
    weight *= 2.2
    new_unit = 'pounds'
elif unit == 'LB' :
    weight /= 2.2
    new_unit = 'kilograms'
else:
    print('wrong unit')
    exit()

print (f'your weight is {round(weight)} {new_unit}')
```
